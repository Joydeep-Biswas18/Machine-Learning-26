### Here the price of Houses depends on multiple Factors like area, bedrooms and age  that is the linear Regression of Multivariate 

```python
df1.info()
# that helps us to find the information of a dataframe from a base level
```

```python
df1['bedrooms'].value_counts
```

```python
df1.bedrooms = df1['bedrooms'].fillna(math.floor(df1['bedrooms'].median()))
# that math.floor(dataframe.median()) that change the null value by their median value
```

```python
df1.bedrooms
```

```python
import sklearn
```

```python
from sklearn import linear_model
# Atfirst we need to call the specific library and then call the model
reg = linear_model.LinearRegression()
```

```python
reg.fit(df1[['Area']],df['price'])
#fit operation helps us to find the most fitted line in the distributed data
```

```python
reg.coef_
# that is the coefficient of the the x that means m
```

```python
reg.predict([[4000]])
```

```python
reg_multi = linear_model.LinearRegression()
```

```python
reg_multi.fit(df1[['Area','bedrooms','age']], df.price)
# always remember when i want to fit then the first attribute always 2-D dataframe
```

## 2nd Example Of Multivarate

```python
second_df = pd.read_csv("Ex-hirring_linear_regression_multivariate.csv")
second_df.head()
```

```python
df2 = second_df.copy()
```

```python
df2.sample(6)
```

```python
df2.info()
```

```python
df2.sample(7)
# sample operation find any rows between anywhere
```

```python
df2.experience  = df2.experience.fillna("zero")
# atfirst we change NAN value by zero so i can apply word2number operation
```

```python
df2.columns=['experience','test_score','interview_score','salary($)']
# that is the method for changing column names
```

```python
df2
```

```python
import word2number
```

```python
!pip install word2number
```

```python
from word2number import w2n
```

```python
df2['test_score'] = df2['test_score'].fillna(math.floor(df2['test_score'].median()))
# replace the nan value by their median value
```

```python
df2.test_score
```

```python
df2.experience = df2.experience.apply(w2n.word_to_num)
# here word2num chamges the string value to digit like "Zero" to 0
```

```python
df2
```

```python
plt.plot(df2['interview_score'],df2['salary($)'])
```

```python
from sklearn import linear_model
```

```python
reg2 = linear_model.LinearRegression()
reg2.fit(df2[['experience','test_score','interview_score']], df2['salary($)'])
```

```python
reg2.coef_
```

```python
reg2.intercept_
```

```python
reg2.predict([[10,6,6]])
```
