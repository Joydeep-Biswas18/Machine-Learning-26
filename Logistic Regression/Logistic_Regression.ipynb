```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
```

```python
df = pd.read_csv("Ex-Logistic_Reg_insurance_data.csv")
df.head()
```

```python
df_copy = df.copy()
```

```python
df_copy.info()
```

```python
df_copy.sample(10)
```

```python
plt.figure(figsize=(8,5))
plt.xlabel("age of People")
plt.ylabel("Innsurance Status")
plt.scatter(df_copy.age , df_copy.bought_insurance, color='green',marker='+')
```

```python
from sklearn.model_selection import train_test_split
x_train ,x_test , y_train ,y_test = train_test_split(df[['age',]] , df['bought_insurance'], test_size=0.1,random_state=42)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=100)
model.fit(x_train , y_train)
```

```python
model.predict(x_test)
```

```python
model.score(x_test,y_test)
```
